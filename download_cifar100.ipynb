```python
import os
import requests
import tarfile

def download_and_extract_cifar100_c(download_url, output_dir):
    # Tạo thư mục lưu trữ nếu chưa tồn tại
    os.makedirs(output_dir, exist_ok=True)
    
    # Đường dẫn lưu file nén
    tar_file_path = os.path.join(output_dir, 'cifar100-c.tar')
    
    # Tải file nén
    print("Đang tải CIFAR-100-C dataset...")
    response = requests.get(download_url, stream=True)
    if response.status_code == 200:
        with open(tar_file_path, 'wb') as f:
            for chunk in response.iter_content(chunk_size=1024):
                if chunk:
                    f.write(chunk)
        print("Tải xuống hoàn tất.")
    else:
        print(f"Không thể tải dataset. Mã lỗi: {response.status_code}")
        return
    
    # Giải nén file tar
    print("Đang giải nén dataset...")
    with tarfile.open(tar_file_path, 'r') as tar:
        tar.extractall(path=output_dir)
    print("Giải nén hoàn tất.")
    
    # Xóa file nén để tiết kiệm không gian
    os.remove(tar_file_path)
    print("Đã xóa file nén.")

# URL của CIFAR-100-C dataset
cifar100_c_url = "https://zenodo.org/record/3555552/files/CIFAR-100-C.tar?download=1"

# Thư mục lưu trữ dataset
output_directory = "./DATA/cifar100-c"

# Tải xuống và giải nén
download_and_extract_cifar100_c(cifar100_c_url, output_directory)
```
